## Programación dinámica

Hay veces en que las soluciones recursivas (o incluso a veces aunque no sean recursivas, las soluciones ingenuas), son extremadamente ineficientes, como en el caso de la implementación recursiva de _Fibonacci_.

Para evitar eso, a veces lo que se hace es calcular una sola vez nomás los resultados del problema que se quiere resolver, y luego guardarlo en un arreglo o tabla.  A eso se le conoce como **Programación dinámica**.

La programación dinámica es una técnica algorítmica que:

* Resuelve problemas dividiéndolos en subproblemas superpuestos.
* Cada subproblema se resuelve una sola vez y se guarda en una tabla de memoización, y de este modo se evita recalcular problemas ya resueltos.

Para aplicar programación dinámica, el algoritmo tiene que poder dividirse en subproblemas superpuestos, y tener una subestructura óptima.

#### Subproblemas superpuestos

* Los mismos subproblemas se resuelven **múltiples veces** durante la ejecución del algoritmo recursivo.

#### Subestructura óptima

* La solución óptima global puede construirse a partir de soluciones óptimas de sus subproblemas.
* Si no hay subestructura óptima, la programación dinámica no garantiza una solución correcta.

### Estructura

### Enfoques de la Programación Dinámica

#### **_Top-down_** (memoización)

* Se resuelve el problema de forma recursiva.
* Parte resolviendo el problema original (el más grande).
* Usa recursión con caché.




#### **_Bottom up_** (tabulación)


* Normalmente es en forma iterativa.
* Se parten resolviendo los subproblemas más pequeños.
* A veces puede calcular subproblemas que no son necesarios.



### Ventajas de la Programación Dinámica

* Evita recálculos innecesarios.
* Mejora la eficiencia cuando hay problemas superpuestos.
* Garantiza la solución óptima.

### Desventajas de la Programación Dinámica

* Mayor complejidad de implementación que soluciones ingenuas.
* Requiere memoria adicional.
* Identificar la estructura óptima puede no ser tan intuitivo.

### ¿Cuándo utilizarla?

* Cuando el problema se puede dividir en subproblemas más pequeños, y superpuestos entre sí.
* Cuando hay una subestructura óptima (cada subproblema resuelto de forma óptima lleva al éxito).
* Cuando se necesita mejorar la eficiencia de una solución recursiva.
* Cuando el problema busca un óptimo (máximo o mínimo).

### Ejemplos notables

#### Problema de la soga

```python
precios = {1:1, 2:5, 3:8, 4:9, 5:10, 6:17, 7:17, 8:20}

def corte_soga(n, precios):
    dp = [0] * (n + 1)
    cortes = [[] for _ in range(n + 1)]

    for largo in range(1, n + 1):
        mejor = -1

        for i in range(1, largo + 1):
            ganancia = precios[i] + dp[largo - i]

            if ganancia > mejor:
                mejor = ganancia
                cortes[largo] = [i] + cortes[largo - i]

        dp[largo] = mejor

    return dp[n], cortes[n], dp


ganancia, cortes, tabla = corte_soga(8, precios)

print("Ganancia máxima:", ganancia)
print("Cortes:", cortes)
print("Tabla DP:", tabla)
```

## Algoritmos Greedy

Greedy es una técnica algorítmica que:

* Construye paso a paso la solución eligiendo la solución que parece mejor en el momento (decisión local).
* Nunca considera tomar otra decisión.
* No explora todas las posibilidades.
* A veces requiere ordenar previamente los datos.

Para que un algoritmo greedy sea correcto debe cumplir con la propiedad de **subestructura óptima** y **propiedad de elección greedy**.  Si no se cumplen no se garantiza el óptimo global.

#### Subestructura óptima

* La solución global se construye a partir de las soluciones óptimas de sus subproblemas.

#### Propiedad de elección greedy

* Se puede tomar la decisión sin necesidad de resolver subproblemas antes.
* Existe siempre una elección local que forma parte de alguna solución óptima.

### Ventajas de los Algoritmos Greedy

* Simples de diseñar e implementar.
* Eficientes (O(n), O(n log n) habitualmente).
* Bajo consumo de memoria.
* Cuando son correctos, dan la solución óptima de una sola pasada.

### Desventajas de los Algoritmos Greedy

* No siempre garantizan el óptimo.
* Una mala elección del criterio greedy lleva a soluciones aceptables pero no óptimas.
* No se pueden deshacer decisiones.

### ¿Cuándo utilizarla?

* Cuando el problema busca un valor óptimo (máximo o mínimo).
* Cuando se cumple la propiedad de elección greedy y subestructura óptima.
* Cuando la solución de programación dinámica es muy costosa y nos basta una buena aproximación, aunque no sea óptima.
